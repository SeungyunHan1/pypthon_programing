```python
# p.11: 전역 변수(global)를 사용하여 이전 계산 결과를 유지하는 함수
result = 0

def adder(num):
    global result # 전역 변수 result를 함수 내에서 사용하겠다고 명시
    result += num
    return result

# 함수 실행 및 결과 확인
print(adder(3))
print(adder(4))
print(adder(5)) # p.11의 실행 결과 예시(12) 확인
```

```python
# p.12: 각각의 결과를 따로 유지하기 위해 두 개의 전역 변수와 함수 정의
result1 = 0
result2 = 0

def adder1(num):
    global result1
    result1 += num
    return result1

def adder2(num):
    global result2
    result2 += num
    return result2

# 각 계산기 함수 실행
print(adder1(3))
print(adder1(4))
print(adder2(3))
print(adder2(7))
```

```python
# p.13: 클래스(Class)라는 틀을 사용하여 여러 개의 객체(인스턴스) 생성
class Calculator:
    def __init__(self):
        self.result = 0 # 각 객체마다 독립적인 result 변수 생성
        
    def adder(self, num):
        self.result += num
        return self.result

# 클래스를 통해 객체(cal1, cal2) 생성
cal1 = Calculator()
cal2 = Calculator()

# 각 객체의 메서드 호출
print(cal1.adder(3))
print(cal1.adder(4))
print(cal2.adder(3))
print(cal2.adder(7))
```

```python
# p.24~26: 데이터 설정(setdata)과 더하기(sum) 기능 구현
class FourCal:
    def setdata(self, first, second):
        self.first = first
        self.second = second
    def sum(self):
        result = self.first + self.second
        return result

a = FourCal()
a.setdata(4, 2)
print("더하기 결과:", a.sum())
```

```python
# p.27: 더하기, 곱하기, 빼기, 나누기 기능이 모두 포함된 클래스
class FourCal:
    def setdata(self, first, second):
        self.first = first
        self.second = second
    def sum(self):
        return self.first + self.second
    def mul(self):
        return self.first * self.second
    def sub(self):
        return self.first - self.second
    def div(self):
        return self.first / self.second

# 두 개의 객체를 이용한 연산 테스트
a = FourCal()
b = FourCal()
a.setdata(4, 2)
b.setdata(3, 7)

print("a sum:", a.sum())
print("a mul:", a.mul())
print("a sub:", a.sub())
print("a div:", a.div())
print("b sub:", b.sub())
```

```python
# p.30: 클래스 없이 작성한 주사위 게임 코드
import random

player1_dice = []
player2_dice = []

for i in range(3):
    player1_dice.append(random.randint(1, 6))
    player2_dice.append(random.randint(1, 6))

print("Player 1 rolled " + str(player1_dice))
print("Player 2 rolled " + str(player2_dice))

if sum(player1_dice) == sum(player2_dice):
    print("Draw")
elif sum(player1_dice) > sum(player2_dice):
    print("Player 1 wins")
else:
    print("Player 2 wins")
```

```python
# p.31: 클래스를 사용하여 구조화한 주사위 게임 코드
from random import randint

class Player:
    def __init__(self):
        self.dice = []
    
    def roll(self):
        self.dice = [] # 현재 주사위 기록 초기화
        for i in range(3):
            self.dice.append(randint(1, 6))
            
    def get_dice(self):
        return self.dice

# 객체 생성 및 게임 진행
player1 = Player()
player2 = Player()

player1.roll()
player2.roll()

print("Player 1 rolled " + str(player1.get_dice()))
print("Player 2 rolled " + str(player2.get_dice()))

# 승패 결정
if sum(player1.get_dice()) == sum(player2.get_dice()):
    print("Draw!")
elif sum(player1.get_dice()) > sum(player2.get_dice()):
    print("Player 1 wins!")
else:
    print("Player 2 wins!")
```
