```python
# p.4: 1부터 5까지의 정수 리스트 만들기
# 1. 기본 방식 (append 사용)
number_list = []
number_list.append(1)
number_list.append(2)
number_list.append(3)
number_list.append(4)
number_list.append(5)
print("직접 append:", number_list)

# 2. for 문을 사용하는 방식
number_list = []
for number in range(1, 6): # 콜론(:) 추가
    number_list.append(number)
print("for 문 사용:", number_list) # 따옴표 추가
```

```python
# p.5~6: List comprehension 방식
# 기본 생성
number_list = [number for number in range(1, 6)]
print("List comprehension:", number_list)

# 함수 적용 (n-1)
minus_list = [number - 1 for number in range(1, 6)]
print("함수 적용(n-1):", minus_list)

# 조건 적용 (홀수만 추출)
a_list = [number for number in range(1, 6) if number % 2 == 1]
print("조건 적용(홀수):", a_list)
```

```python
# p.7: 문자열 처리 및 튜플 저장
sentence = ['I', 'Love', 'Python', 'Soooooo', 'MUCH!!!']

# 소문자 변환
lower_words = [word.lower() for word in sentence]
print("소문자 변환:", lower_words)

# 길이가 6보다 큰 단어 추출
long_words = [word for word in sentence if len(word) > 6]
print("길이 6 초과 단어:", long_words)

# 제곱/세제곱 튜플 생성
power_list = [(x, x ** 2, x ** 3) for x in range(10)]
print("제곱/세제곱 튜플 (첫 3개):", power_list[:3])
```

```python
# p.8: 중첩 for 문을 리스트 내포로 구현
rows = range(1, 4)
cols = range(1, 3)

cells = [(row, col) for row in rows for col in cols]
for cell in cells: # 콜론(:) 추가
    print(cell)

# 연습 예제: j가 i보다 작은 경우
practice = [(i, j) for i in range(5) for j in range(i)]
print("연습 예제 결과:", practice)
```

```python
# p.9~12: 딕셔너리 및 셋 내포
word = 'letters'

# 딕셔너리 내포: 글자별 빈도수 (효율적인 set 활용 방식)
letter_counts = {letter: word.count(letter) for letter in set(word)}
print("딕셔너리 내포(set 활용):", letter_counts)

# Set 내포: 3으로 나눈 나머지가 1인 수
a_set = {number for number in range(1, 6) if number % 3 == 1}
print("Set 내포 결과:", a_set)
```

```python
# p.14~15: 덧셈 함수 정의
def sum_func(a, b): # 콜론(:) 추가
    return a + b

# 다양한 타입으로 호출
print("정수 합:", sum_func(1, 2))
print("실수 합:", sum_func(1.3, 3.1))
print("문자열 합:", sum_func('love ', 'python'))
```

```python
# p.16~17: zip 함수 활용
days = ['Monday', 'Tuesday', 'Wednesday']
fruits = ['banana', 'orange', 'peach']
drinks = ['coffee', 'tea', 'beer']
desserts = ['tiramisu', 'ice cream', 'pie', 'pudding']

# f-string 수정 및 콜론 추가
for day, fruit, drink, dessert in zip(days, fruits, drinks, desserts):
    print(f"{day} : drink {drink} - eat {fruit} - enjoy {dessert}")

# 영어-프랑스 사전 만들기
english = 'Monday', 'Tuesday', 'Wednesday'
french = 'Lundi', 'Mardi', 'Mercredi'
print("영어-불어 사전:", dict(zip(english, french)))
```

```python
# p.18~19: 인자 처리 방식
def menu(wine, entree, dessert):
    return {'wine': wine, 'entree': entree, 'dessert': dessert}

print("위치 인자:", menu('chardonnay', 'chicken', 'cake'))
print("키워드 인자:", menu(entree='beef', dessert='bagel', wine='bordeaux'))
print("인자 혼용:", menu('frontenac', dessert='flan', entree='fish'))
```

```python
# p.20~21: 기본값 설정
def menu_default(wine, entree, dessert='pudding'):
    return {'wine': wine, 'entree': entree, 'dessert': dessert}

print("기본값 사용:", menu_default('chardonnay', 'chicken'))

# 복합 인자 예제
def menu_price(price, wine='chardonnay', entree='chicken', dessert='pudding'):
    return {'price': price, 'wine': wine, 'entree': entree, 'dessert': dessert}

print("다양한 인자 호출:", menu_price(100, entree='beef'))
```
