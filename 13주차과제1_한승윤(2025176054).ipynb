```python
menu_xml = '''<?xml version="1.0"?>
<menu>
    <breakfast hours="7-11">
        <item price="$6.00">breakfast burritos</item>
        <item price="$4.00">pancakes</item>
    </breakfast>
    <lunch hours="11-3">
        <item price="$5.00">hamburger</item>
    </lunch>
    <dinner hours="3-10">
        <item price="8.00">spaghetti</item>
    </dinner>
</menu>
'''
with open('menu.xml', 'w', encoding='utf-8') as f:
    f.write(menu_xml)

print("menu.xml 생성 완료")
```

```python
import xml.etree.ElementTree as et

tree = et.ElementTree(file='menu.xml')
root = tree.getroot()

print(root.tag)
print(type(root))
```

```python
# Cell 2: root의 자식 / 손자 태그와 속성 출력
for child in root:
    print('tag:', child.tag, 'attributes:', child.attrib)
    for grandchild in child:
        print('\ttag:', grandchild.tag, 'attributes:', grandchild.attrib)
```

```python
# Cell 3: menu 하위 태그 수와 breakfast의 item 수
print("len(root):", len(root))         # 3
print("len(root[0]):", len(root[0]))   # 2
print("type(root[0]):", type(root[0]))
```

```python
# Cell 4: 특정 태그의 속성 값 접근 - get, keys, items
print(root[0])                    # breakfast
print(root[0].get("hours"))       # '7-11'
print(root[0].keys())             # ['hours']
print(root[0].items())            # [('hours', '7-11')]
```

```python
# Cell 5: find로 태그 이름으로 찾기
lunch = root.find("lunch")
print(lunch)
print(lunch.get("hours"))
print(lunch.keys())
print(lunch.items())
print(type(lunch))
```

```python
# Cell 6: findall + .text
print(root[0].findall('item'))

for item in root[0].findall('item'):
    print(item.get('price'))
    print(item.text)
```

```python
# Cell 7: json.dumps - 딕셔너리를 JSON 문자열로
import json

j1 = {"name":"홍길동", "birth":"0525", "age": 30}
print(j1)
print(json.dumps(j1))
```

```python
# Cell 8: json.dumps - indent, list, tuple
print(json.dumps(j1, indent=2))
print(json.dumps([1, 2, 3]))
print(json.dumps((4, 5, 6)))
```

```python
# Cell 9: json.loads - JSON 문자열을 파이썬 객체로
j1 = {"name":"홍길동", "birth":"0525", "age": 30}
d1 = json.dumps(j1)
print(json.loads(d1))
```

```python
# Cell 13: BeautifulSoup으로 h1 태그 가져오기
from urllib.request import urlopen
from bs4 import BeautifulSoup

html = urlopen("http://pythonscraping.com/pages/page1.html")
bsObj = BeautifulSoup(html.read(), "html.parser")
print(bsObj.h1)
```

```python
# Cell 14: BeautifulSoup - 계층적 태그 접근
print(bsObj.html.prettify())

print(bsObj.h1)
print(bsObj.html.body.h1)
print(bsObj.body.h1)
print(bsObj.html.h1)
```
